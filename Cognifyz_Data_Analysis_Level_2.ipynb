## ⭐ Task 1: Restaurant Ratings
> **Objective:** Analyze the distribution of aggregate ratings, find the most common rating range, and calculate the average number of votes received by restaurants.

```python
# Most common rating (Mode) and average votes
common_rating = df['Aggregate rating'].mode()[0]
avg_votes = df['Votes'].mean()

print(f" Most Common Aggregate Rating: {common_rating}")
print(f" Average Number of Votes per Restaurant: {avg_votes:.2f}\n")
```

```python
# Visualizing Aggregate Ratings Distribution (Histogram)
plt.figure(figsize=(8, 4))
plt.hist(df['Aggregate rating'], bins=20, color='orange', edgecolor='black')
plt.title('Distribution of Aggregate Ratings', fontsize=13, fontweight='bold')
plt.xlabel('Aggregate Rating', fontsize=11)
plt.ylabel('Frequency (Number of Restaurants)', fontsize=11)
plt.tight_layout()
plt.show()
```

## 🍽️ Task 2: Cuisine Combination
> **Objective:** Identify the most common combinations of cuisines in the dataset and determine whether certain combinations tend to receive higher average ratings.

```python
# common cuisine combinations
cuisine_combinations = df['Cuisines'].dropna().value_counts().head(10)
print("🏆 Top 10 Most Common Cuisine Combinations:")
for combo, count in cuisine_combinations.items():
    print(f"• {combo}: {count} restaurants")
```

```python
# top combinations average rating
cuisine_rating = df[['Cuisines', 'Aggregate rating']].dropna()
avg_rating_per_combo = cuisine_rating.groupby('Cuisines')['Aggregate rating'].mean()

print("\nTop 5 Highest Rated Cuisine Combinations (with at least 5 restaurants):")
combo_counts = cuisine_rating['Cuisines'].value_counts()
valid_combos = combo_counts[combo_counts >= 5].index
filtered_avg_rating = avg_rating_per_combo[valid_combos]
print(filtered_avg_rating.sort_values(ascending=False).head(5).round(2))
```

## 🗺️ Task 3: Geographic Analysis
> **Objective:** Plot the locations of restaurants on a map using their latitude and longitude coordinates to identify geographic clustering and concentration patterns.

```python
# Confirm the Latitude and Longitude columns cleaned or not
geo_df = df[['Latitude', 'Longitude', 'City', 'Aggregate rating']].dropna()
print(f"📊 Total valid geographic points found: {len(geo_df)}")
```

```python
# Plotting restaurant locations on a scatter plot (Latitude vs Longitude)
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x='Longitude',
    y='Latitude',
    data=geo_df,
    hue='City',
    palette='viridis',
    s=20,
    alpha=0.7,
    legend=False
)
plt.title('Geographic Distribution of Restaurants', fontsize=14, fontweight='bold')
plt.xlabel('Longitude', fontsize=12)
plt.ylabel('Latitude', fontsize=12)
plt.tight_layout()
plt.show()
```

## 🔗 Task 4: Restaurant Chains
> **Objective:** Identify if there are any restaurant chains present in the dataset and analyze the ratings and popularity of different restaurant chains.

```python
# Restaurant chains identify karna
chain_counts = df['Restaurant Name'].value_counts()
chains = chain_counts[chain_counts > 1]
print(f"Total Unique Restaurants: {df['Restaurant Name'].nunique()}")
print(f"Total Restaurant Chains (appearing > 1 time): {len(chains)}")
print("\nTop 5 Restaurant Chains by Outlets:")
print(chains.head(5))
```

```python
# Top restaurant chains ki ratings aur popularity (votes) analyze karna
chain_data = df[df['Restaurant Name'].isin(chains.index)]
chain_analysis = chain_data.groupby('Restaurant Name').agg({
    'Aggregate rating': 'mean',
    'Votes': 'sum',
    'Restaurant ID': 'count'
}).rename(columns={'Restaurant ID': 'Number of Outlets'}).sort_values(by='Number of Outlets', ascending=False)

print("\n📊 Top Chains Analysis (Outlets, Avg Rating, Total Votes):")
print(chain_analysis.head(10))
```

```python
# Visualization: Top 10 Restaurant Chains by Number of Outlets
plt.figure(figsize=(12, 6))
top_chains = chain_analysis.head(10)
sns.barplot(x=top_chains.index, y=top_chains['Number of Outlets'], palette='viridis')
plt.title('Top 10 Restaurant Chains by Number of Outlets', fontsize=14, fontweight='bold')
plt.xlabel('Restaurant Chain', fontsize=12)
plt.ylabel('Number of Outlets', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
