## 🔍 Task 1: Top Cuisines
> **Objective:** Determine the top three most common cuisines in the dataset and calculate the percentage of restaurants that serve each of them.

```python
# Top 3 Cuisines
cuisine_counts = df['Cuisines'].dropna().value_counts()
top_3_cuisines = cuisine_counts.head(3)
```

```python
# Percentage calculate
total_restaurants = len(df['Restaurant ID'])
total_restaurants
```

```python
top_3_percentage = (top_3_cuisines / total_restaurants) * 100
```

```python
print("--- Top 3 Most Common Cuisines ---")
print(top_3_cuisines)

print("\n--- Percentage of Total Restaurants ---")
print(top_3_percentage.round(2).astype(str) + '%')
```

## 🏙️ Task 2: City Analysis
> **Objective:** Identify the city with the highest number of restaurants in the dataset, calculate the average rating for restaurants in each city, and determine the city with the highest average rating.

```python
# Identify the city with the highest number of restaurants
city_counts = df['City'].value_counts()
top_city = city_counts.idxmax()
top_city_count = city_counts.max()
print(f"🏙️ City with the Highest Number of Restaurants:")
print(f"• {top_city} ({top_city_count} restaurants)\n")
```

```python
# Calculate the average rating for restaurants in each city
avg_rating_per_city = df.groupby('City')['Aggregate rating'].mean()
```

```python
# Determine the city with the highest average rating
highest_rating_city = avg_rating_per_city.idxmax()
highest_rating_value = avg_rating_per_city.max()
print(f"⭐ City with the Highest Average Rating:")
print(f"• {highest_rating_city} (Average Rating: {highest_rating_value:.2f})")
```

```python
print("\n------------------------------------------")
print("📊 Top 5 Cities by Average Rating:")
print(avg_rating_per_city.sort_values(ascending=False).head(5).round(2))
print("==========================================\n")
```

## Task 3: Price Range Distribution
> **Objective:** Analyze the distribution of price ranges among the restaurants and calculate the percentage of restaurants in each price range category.

```python
# Price range counts and percentages
price_counts = df['Price range'].value_counts().sort_index()
price_percentage = (price_counts / len(df)) * 100
```

```python
print("📊 Price Range:")
for pr, count in price_counts.items():
    pct = price_percentage[pr]
    print(f"Price Range {pr}: {count} restaurants ({pct:.2f}%)")
```

```python
# Visualizing the Price Range Distribution using a Bar Chart
plt.figure(figsize=(7, 4))
price_counts.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Distribution of Restaurant Price Ranges', fontsize=13, fontweight='bold')
plt.xlabel('Price Range (1: Low, 4: High)', fontsize=11)
plt.ylabel('Number of Restaurants', fontsize=11)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Task 4: Online Delivery Analysis
> **Objective:** Calculate the percentage of restaurants that offer online delivery and compare the average ratings between restaurants with and without online delivery.

```python
# Online delivery counts and percentage
delivery_counts = df['Has Online delivery'].value_counts()
delivery_percentage = (delivery_counts / len(df)) * 100
```

```python
print("📊 Online Delivery Availability:")
for status, count in delivery_counts.items():
    pct = delivery_percentage[status]
    print(f"Delivery '{status}': {count} restaurants ({pct:.2f}%)")
```

```python
# Online delivery vs bina delivery walo ki average rating compare karna
avg_rating_delivery = df.groupby('Has Online delivery')['Aggregate rating'].mean()
avg_rating_delivery
```

```python
print("\n Average Rating Comparison:")
for status, avg_rat in avg_rating_delivery.items():
    print(f"• Delivery '{status}': Average Rating = {avg_rat:.2f}")
```
